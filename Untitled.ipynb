```python
import os
import json
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import joblib

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    mean_absolute_error, mean_squared_error, r2_score,
)

sns.set_theme(style="whitegrid", palette="viridis")

# --- Locate the CSV automatically, or ask you to pick it ---
FILENAME = "legal_cases_five_tasks.csv"

def find_data_file(filename):
    # 1. check current working directory
    if os.path.exists(filename):
        return os.path.abspath(filename)
    # 2. check common folders
    candidates = [
        os.path.join(os.path.expanduser("~"), "Downloads", filename),
        os.path.join(os.path.expanduser("~"), "Desktop", filename),
        os.path.join(os.path.expanduser("~"), "Documents", filename),
    ]
    for c in candidates:
        if os.path.exists(c):
            return c
    # 3. fall back to a file picker dialog
    try:
        from tkinter import Tk
        from tkinter.filedialog import askopenfilename
        Tk().withdraw()
        path = askopenfilename(title=f"Select {filename}", filetypes=[("CSV files", "*.csv")])
        if path:
            return path
    except Exception:
        pass
    raise FileNotFoundError(
        f"Could not find '{filename}'. Set DATA_PATH manually below, e.g.\n"
        r'DATA_PATH = r"C:\Users\YourName\Downloads\legal_cases_five_tasks.csv"'
    )

DATA_PATH = find_data_file(FILENAME)
print("Using data file:", DATA_PATH)

MODELS_DIR = "models"
PLOTS_DIR = "plots"
os.makedirs(MODELS_DIR, exist_ok=True)
os.makedirs(PLOTS_DIR, exist_ok=True)

RANDOM_STATE = 42
TEST_SIZE = 0.2
```

```python
df = pd.read_csv(DATA_PATH)
print(df.shape)
df = df.drop_duplicates().reset_index(drop=True)

for col in ["Cross_Jurisdictional", "High_Quality_Law_Firms"]:
    if df[col].dtype == object:
        df[col] = df[col].astype(str).str.strip().str.lower().map({"true": True, "false": False})

for col in df.columns:
    if df[col].isnull().sum() == 0:
        continue
    if pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode().iloc[0])

df.head()
```

```python
def save(fig, name):
    fig.savefig(os.path.join(PLOTS_DIR, name), dpi=150, bbox_inches="tight")
    plt.show()

fig, ax = plt.subplots(figsize=(8, 5))
sns.countplot(data=df, x="Case_Type", order=df["Case_Type"].value_counts().index, ax=ax)
ax.set_title("Case Type Distribution"); ax.tick_params(axis="x", rotation=30)
save(fig, "01_case_type_distribution.png")

fig, ax = plt.subplots(figsize=(9, 5))
sns.countplot(data=df, x="Case_Type", hue="Outcome", ax=ax)
ax.set_title("Outcome by Case Type"); ax.tick_params(axis="x", rotation=30)
save(fig, "02_outcome_by_case_type.png")

fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(df["Penalty_Prediction"], kde=True, bins=40, ax=ax)
ax.set_title("Penalty Amount Distribution")
save(fig, "03_penalty_distribution.png")

fig, ax = plt.subplots(figsize=(16, 13))
sns.heatmap(df.select_dtypes(include=["number"]).corr(), cmap="coolwarm", center=0, ax=ax)
ax.set_title("Correlation Heatmap — Numeric Features")
save(fig, "04_correlation_heatmap.png")

fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="Jurisdiction", y="Liability_Score", ax=ax)
ax.set_title("Liability Score by Jurisdiction")
save(fig, "05_liability_by_jurisdiction.png")

fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(data=df, x="Financial_Calculation", y="Penalty_Prediction", hue="Case_Type", alpha=0.6, ax=ax)
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_title("Financial Calculation vs Penalty Prediction")
save(fig, "06_financial_vs_penalty.png")

fig, ax = plt.subplots(figsize=(9, 5))
year_counts = df.groupby(["Case_Year", "Outcome"]).size().reset_index(name="count")
sns.lineplot(data=year_counts, x="Case_Year", y="count", hue="Outcome", marker="o", ax=ax)
ax.set_title("Case Volume by Year and Outcome")
save(fig, "07_case_year_trend.png")
```

```python
CATEGORICAL_COLS = ["Jurisdiction", "Case_Type", "Outcome", "Cross_Jurisdictional",
                     "High_Quality_Law_Firms", "Article_Prediction", "Cause_Prediction"]

encoders = {}
df_enc = df.copy()
for col in CATEGORICAL_COLS:
    le = LabelEncoder()
    df_enc[col] = le.fit_transform(df_enc[col].astype(str))
    encoders[col] = le

joblib.dump(encoders, os.path.join(MODELS_DIR, "label_encoders.joblib"))

DROP_COLS = ["Case_ID", "Case_Analysis"]
df_enc.head()
```

```python
TASKS = {
    "case_type": {"type": "classification", "target": "Case_Type",
                  "exclude": ["Article_Prediction", "Cause_Prediction"]},
    "outcome": {"type": "classification", "target": "Outcome", "exclude": []},
    "article_prediction": {"type": "classification", "target": "Article_Prediction",
                            "exclude": ["Case_Type", "Cause_Prediction"]},
    "penalty_prediction": {"type": "regression", "target": "Penalty_Prediction", "exclude": []},
    "liability_score": {"type": "regression", "target": "Liability_Score", "exclude": []},
}

def build_Xy(df_enc, target, exclude):
    drop = set(DROP_COLS + exclude + [target])
    feature_cols = [c for c in df_enc.columns if c not in drop]
    return df_enc[feature_cols], df_enc[target], feature_cols
```

```python
all_metrics = {}

for task_name, cfg in TASKS.items():
    print(f"--- {task_name} ({cfg['type']}) ---")
    X, y, feature_cols = build_Xy(df_enc, cfg["target"], cfg["exclude"])
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE,
        stratify=y if cfg["type"] == "classification" else None,
    )

    if cfg["type"] == "classification":
        model = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE,
                                        n_jobs=-1, class_weight="balanced")
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred, average="weighted")
        all_metrics[task_name] = {"accuracy": acc, "f1_weighted": f1}
        print(f"  accuracy={acc:.3f}  f1={f1:.3f}")

        labels = encoders[cfg["target"]].classes_
        cm = confusion_matrix(y_test, y_pred)
        fig, ax = plt.subplots(figsize=(6, 5))
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=labels, yticklabels=labels, ax=ax)
        ax.set_xlabel("Predicted"); ax.set_ylabel("Actual"); ax.set_title(f"Confusion Matrix — {task_name}")
        plt.xticks(rotation=30, ha="right")
        save(fig, f"cm_{task_name}.png")

    else:
        model = RandomForestRegressor(n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        mae = mean_absolute_error(y_test, y_pred)
        rmse = mean_squared_error(y_test, y_pred) ** 0.5
        r2 = r2_score(y_test, y_pred)
        all_metrics[task_name] = {"mae": mae, "rmse": rmse, "r2": r2}
        print(f"  MAE={mae:.3f}  RMSE={rmse:.3f}  R2={r2:.3f}")

        fig, ax = plt.subplots(figsize=(6, 5))
        sns.scatterplot(x=y_test, y=y_pred, alpha=0.5, ax=ax)
        lims = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
        ax.plot(lims, lims, "r--", linewidth=1)
        ax.set_xlabel("Actual"); ax.set_ylabel("Predicted"); ax.set_title(f"Predicted vs Actual — {task_name}")
        save(fig, f"fit_{task_name}.png")

    importances = model.feature_importances_
    order = np.argsort(importances)[::-1][:15]
    fig, ax = plt.subplots(figsize=(8, 6))
    sns.barplot(x=importances[order], y=np.array(feature_cols)[order], ax=ax)
    ax.set_title(f"Top Feature Importances — {task_name}")
    save(fig, f"featimp_{task_name}.png")

    joblib.dump({"model": model, "feature_columns": feature_cols},
                os.path.join(MODELS_DIR, f"{task_name}_model.joblib"))
    print(f"  saved -> {MODELS_DIR}/{task_name}_model.joblib\n")

with open(os.path.join(MODELS_DIR, "training_metrics.json"), "w") as f:
    json.dump(all_metrics, f, indent=2, default=str)

pd.DataFrame(all_metrics).T
```

```python
def predict(task_name, record: dict):
    bundle = joblib.load(os.path.join(MODELS_DIR, f"{task_name}_model.joblib"))
    model, feature_cols = bundle["model"], bundle["feature_columns"]

    row = pd.DataFrame([record])
    for col in CATEGORICAL_COLS:
        if col in row.columns:
            le = encoders[col]
            known = set(le.classes_)
            row[col] = row[col].astype(str).apply(lambda v: v if v in known else le.classes_[0])
            row[col] = le.transform(row[col])

    X = row[feature_cols]
    pred = model.predict(X)[0]

    target = TASKS[task_name]["target"]
    if target in encoders:
        pred = encoders[target].inverse_transform([int(round(pred))])[0]
    return pred

sample = df.iloc[0].to_dict()
print("case_type ->", predict("case_type", sample))
print("penalty_prediction ->", predict("penalty_prediction", sample))
```
